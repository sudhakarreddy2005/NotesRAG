```python
# create a simple text file
import os
os.makedirs("../data/text_files" , exist_ok = True)
```

```python
simpletextfiles = {
    "../data/text_files/python_intro.txt": """python pogmasdfsfdsdfasjdfjshfskdjhfskfhsdhfsfsdkfsdfsfafsdfsfassfsfsd""",
    "../data/text_files/machinelearning.txt" : """ ml is subset of ai asfhsdfsfasdfsDFASFSDFSFSFSF"""
}
```

```python
for filepath , content in simpletextfiles.items():
    with open(filepath , "w") as f:
        f.write(content)
print("simple text files are created")
        
```

```python
# textloader

# from langchain.document_loaders import TextLoader
from langchain_community.document_loaders import TextLoader
loader = TextLoader("../data/text_files/python_intro.txt" , encoding = 'utf-8')
print("loader" , loader)
documents = loader.load()
print("content is " ,documents)
```

```python
# Directory loader  is for loading of all files at a time
from langchain_community.document_loaders import DirectoryLoader
dir_loader = DirectoryLoader(
    "../data/text_files" ,  #locationpath 
    glob = "**/*.txt" , # pattern for matchc
    loader_cls = TextLoader, #loader class to use
    loader_kwargs = {'encoding' : 'utf-8'}
)
```

```python
print(dir_loader)
```

```python
documents = dir_loader.load()
print(documents)
```

```python
# now  pdf loader
from langchain_community.document_loaders import PyPDFLoader , PyMuPDFLoader
dir_loader = DirectoryLoader(
    "../data/pdf" , 
    glob = "**/*.pdf" , 
    loader_cls  = PyMuPDFLoader,
    
)
```

```python
pdf_doc_loader = dir_loader.load()
print(pdf_doc_loader)
```
